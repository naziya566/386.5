```python
x = np.linspace(0, 2 * np.pi, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.tan(x)
y4 = np.exp(x)

fig, axes = plt.subplots(2, 2)

axes[0, 0].plot(x, y1)
axes[0, 0].set_title('Sin Function')

axes[0, 1].plot(x, y2)
axes[0, 1].set_title('Cosine Function')

axes[1, 0].plot(x, y3)
axes[1, 0].set_title('Tangent Function')

axes[1, 1].plot(x, y4)
axes[1, 1].set_title('Exponential Function')

plt.tight_layout()
plt.show()
fig,axes=plt.subplots(2,2)
axes[0,0].plot(x,y1)
axes[0,1].plot(x,y2)
axes[1,0].plot(x,y3)
axes[1,1].plot(x,y4)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate data
x = np.linspace(0, 2 * np.pi, 200) # Increased to 200 for a smoother tangent line
y1 = np.sin(x)
y2 = np.cos(x)

# Clean up tangent asymptotes to avoid vertical bridging lines
y3 = np.tan(x)
y3[abs(y3) > 10] = np.nan 

y4 = np.exp(x)

# Create a single figure and axes
plt.figure(figsize=(10, 6))

# Plot all lines sequentially on the same plot
plt.plot(x, y1, label='Sin Function', linewidth=2)
plt.plot(x, y2, label='Cosine Function', linewidth=2)
plt.plot(x, y3, label='Tangent Function', linewidth=2)
plt.plot(x, y4, label='Exponential Function', linewidth=2)

# Set labels, title, and limits
plt.title('Trigonometric & Exponential Functions')
plt.xlabel('X values')
plt.ylabel('Y values')
plt.ylim(-2.5, 15) # Limits y-axis so sin/cos remain visible

# Add a legend to distinguish the lines
plt.legend()
plt.grid(True)

# Display the plot
plt.show()
```
